# Cumulative Distance Problems

## Part 1: Cumulative Distance

### Problem
You are given a sequence of distances. At each step, compute the total distance traveled so far and store it in a new list.

Write a function `cumulative_distance(distances)` that returns one running total for each position in the input.

### Example 1
Input: `[2, 4, 1, 7]`
Output: `[2, 6, 7, 14]`

### Explanation 1
- Start with a total of 0
- Add `2` -> total is `2`
- Add `4` -> total is `6`
- Add `1` -> total is `7`
- Add `7` -> total is `14`

### Example 2
Input: `[3, 5, 2, 6]`
Output: `[3, 8, 10, 16]`

### Explanation 2
- `3` becomes the first running total
- `5` is added to the previous total: `3 + 5 = 8`
- `2` is added: `8 + 2 = 10`
- `6` is added: `10 + 6 = 16`

### Requirements
- Return a list of cumulative totals.
- Do not modify the input list.
- The function should work for any list length.

### Bonus
Try to solve this without creating a second list for the running totals.

```python
def cumulative_distance(distances):
    """Return the running total distance after each segment."""
    output = []
    for i in distances:
        if not output:
            output.append(i)
        else:
            output.append(output[-1] + i)
    return output


# Example test case
print(cumulative_distance([2, 4, 1, 7]))
# Expected: [2, 6, 7, 14]
print(cumulative_distance([1, 1, 1, 1]))
# Expected: [1, 2, 3, 4]
print(cumulative_distance([5, 10, 15, 0]))
# Expected: [5, 15, 30, 30]
```

## Part 2: Same-Value Distance

### Problem
For each index, find every other index with the same value. Compute the absolute difference between the current index and each matching index, then add those differences together.

Write a function `cumulative_distance_2(values)` that returns one result for each position in the list.

### Example 1
Input: `[4, 4, 1, 4, 3, 2, 1]`
Output: `[4, 3, 4, 5, 0, 0, 4]`

### Explanation 1
At index `0`, the value `4` appears at positions `0`, `1`, and `3`.
The distances are `|0 - 0| = 0`, `|0 - 1| = 1`, and `|0 - 3| = 3`, so the total is `4`.
At index `1`, the matching positions are `0`, `1`, and `3`, which gives `1 + 0 + 2 = 3`.
At index `3`, the matching positions give `3 + 2 + 0 = 5`.

### Example 2
Input: `[5, 4, 5, 3, 4]`
Output: `[2, 3, 2, 0, 3]`

### Explanation 2
- The value `5` appears at indices `0` and `2`, so both positions get `|0 - 2| = 2`.
- The value `4` appears at indices `1` and `4`, so both positions get `|1 - 4| = 3`.
- The value `3` appears only once, so its contribution is `0`.

### Requirements
- Return one value per position.
- Consider all matching positions with the same value.
- Keep the original list unchanged.
- Handle lists of any length.

### Challenge
Try to solve this efficiently for larger inputs without checking every pair one by one.

```python
def cumulative_distance_2(values):
    """Return the summed distance from each index to all matching values."""
    result = []
    for i in range(len(values)):
        total = 0
        for j in range(len(values)):
            if values[j] == values[i]:
                total += abs(i - j)
        result.append(total)
    return result


# Example test cases
print(cumulative_distance_2([4, 4, 1, 4, 3, 2, 1]))
# Expected: [4, 3, 4, 5, 0, 0, 4]

print(cumulative_distance_2([5, 4, 5, 3, 4]))
# Expected: [2, 3, 2, 0, 3]
```

## Part 3: Nearest Same-Value Distance

### Problem
For each index, find the nearest index with the same value. Return the distance to that nearest match.

If a value appears only once, return `0`.

Write a function `nearest_same_distance(values)`.

### Example 1
Input: `[1, 2, 1, 3, 3]`
Output: `[2, 0, 2, 1, 1]`

### Explanation 1
- The value `1` appears at indices `0` and `2`, so each gets distance `2`.
- The value `2` appears only once, so its distance is `0`.
- The value `3` appears at indices `3` and `4`, so each gets distance `1`.

### Example 2
Input: `[7, 4, 7, 2, 2, 9]`
Output: `[2, 0, 2, 1, 1, 0]`

### Explanation 2
- The `7` values are 2 positions apart, so each gets distance `2`.
- The value `4` appears only once, so its distance is `0`.
- The `2` values are adjacent, so each gets distance `1`.
- The value `9` appears only once, so its distance is `0`.

### Requirements
- Return one value per position.
- Use the closest matching index only.
- Keep the original list unchanged.

```python
def nearest_same_distance(values):
    """Return the distance to the nearest index with the same value."""
    nearest = [None] * len(values)
    last_seen = {}

    for i, value in enumerate(values):
        if value in last_seen:
            nearest[i] = i - last_seen[value]
        last_seen[value] = i

    next_seen = {}
    for i in range(len(values) - 1, -1, -1):
        value = values[i]
        if value in next_seen:
            distance = next_seen[value] - i
            if nearest[i] is None or distance < nearest[i]:
                nearest[i] = distance
        next_seen[value] = i

    return [distance if distance is not None else 0 for distance in nearest]


# Example test cases
print(nearest_same_distance([1, 2, 1, 3, 3]))
# Expected: [2, 0, 2, 1, 1]

print(nearest_same_distance([7, 4, 7, 2, 2, 9]))
# Expected: [2, 0, 2, 1, 1, 0]
```

## Part 4: Farthest Same-Value Distance

### Problem
For each position, find the farthest position containing the same value. Return the distance to that position. If the value appears only once, return `0`.

Write a function `farthest_same_distance(values)` that returns one distance for each position.

### Example 1
Input: `[4, 1, 4, 4, 2, 1]`
Output: `[3, 4, 2, 3, 0, 4]`

### Explanation 1
- The `4`s are at indices `0`, `2`, and `3`. Their farthest-match distances are `3`, `2`, and `3`.
- The `1`s are at indices `1` and `5`, so each gets distance `4`.
- The `2` appears only once, so its distance is `0`.

### Example 2
Input: `[8, 6, 8, 6, 8]`
Output: `[4, 2, 2, 2, 4]`

### Explanation 2
- The `8`s are at indices `0`, `2`, and `4`, so their farthest-match distances are `4`, `2`, and `4`.
- The `6`s are at indices `1` and `3`, so each gets distance `2`.

### Requirements
- Return one value per position.
- Use the farthest matching position, not the sum of distances.
- Keep the original list unchanged.

```python
def farthest_same_distance(values):
    """Return the distance to the farthest index with the same value."""
    first_seen = {}
    last_seen = {}

    for i, value in enumerate(values):
        if value not in first_seen:
            first_seen[value] = i
        last_seen[value] = i

    return [
        max(i - first_seen[value], last_seen[value] - i)
        for i, value in enumerate(values)
    ]


# Example test cases
print(farthest_same_distance([4, 1, 4, 4, 2, 1]))
# Expected: [3, 4, 2, 3, 0, 4]

print(farthest_same_distance([8, 6, 8, 6, 8]))
# Expected: [4, 2, 2, 2, 4]
```
